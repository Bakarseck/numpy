```python
generator = np.random.default_rng(123)
grades = np.round(generator.uniform(low=0.0, high=10.0, size=(10, 2)))
grades[[1, 2, 5, 7], 0] = np.nan

# Si la première note manque, prendre la deuxième
note_retenue = np.where(
    np.isnan(grades[:, 0]),
    grades[:, 1],
    grades[:, 0]
)

# Ajouter les notes retenues comme troisième colonne
grades = np.column_stack((grades, note_retenue))

print(grades)
```

```python
url = (
   "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
)

original = np.genfromtxt(
    url,
    delimiter=";",
    dtype=np.float64
)

wine = np.genfromtxt(url, delimiter=";", dtype=np.float32)

print("Mémoire :", wine.nbytes, "octets")  # 7680

difference = np.nansum(
    np.abs(original - wine.astype(np.float64))
)# 0

print("Différence totale :", difference)
print("Précision respectée :", difference < 1e-3)

wine = wine[1:]

print("Colonnes entièrement NaN :", np.isnan(wine).all(axis=0))

ph = wine[:, 8]
sulphates = wine[:, 9]
alcool = wine[:, 10]
qualite = wine[:, 11]

# 3. Au moins un vin avec plus de 20 % d'alcool ?
print("Alcool > 20 % :", bool(np.any(alcool > 20)))

# 4. Pourcentage moyen d'alcool
print("Alcool moyen :", np.nanmean(alcool))


# 5. Statistiques du pH
print("pH minimum :", np.nanmin(ph))
print("pH maximum :", np.nanmax(ph))
print("Quartiles du pH :", np.nanpercentile(ph, [25, 50, 75]))
print("pH médian :", np.nanmedian(ph))

# 6. Qualité moyenne des vins sous le seuil du 20e percentile
seuil = np.nanpercentile(sulphates, 20)
print(
    "Qualité moyenne, faibles sulphates :",
    np.nanmean(qualite[sulphates <= seuil])
)

# 7. Moyenne de chaque colonne pour la meilleure/pire qualité
meilleure = np.nanmax(qualite)
pire = np.nanmin(qualite)

print("Meilleure qualité :", meilleure)
print(np.nanmean(wine[qualite == meilleure], axis=0))

print("Pire qualité :", pire)
print(np.nanmean(wine[qualite == pire], axis=0))
```

```python
import numpy as np
from itertools import permutations

# Charger la matrice des écarts prédits
forecasts = np.loadtxt("https://raw.githubusercontent.com/01-edu/Branch-AI/refs/heads/master/piscine/week01/day01/ex09/data/model_forecasts.txt")

# Tous les ordres possibles des 10 équipes
tirages = np.array(list(permutations(range(10))), dtype=np.int8)

# Chaque ligne devient 5 matchs, avec 2 équipes par match
matchs = tirages.reshape(-1, 5, 2)

# Récupérer les écarts prédits pour chaque match
ecarts = forecasts[matchs[:, :, 0], matchs[:, :, 1]]

# Calculer le coût total de chaque tirage
couts = np.sum(ecarts ** 2, axis=1)

# Garder le tirage qui minimise ce coût
meilleur = matchs[np.argmin(couts)]

# Format demandé : 2 lignes et 5 colonnes
print(meilleur.T)
```
